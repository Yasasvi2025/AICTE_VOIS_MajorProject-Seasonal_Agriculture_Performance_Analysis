# **6. Data Cleaning & Preprocessing**
### **Purpose**
Prepare the dataset for reliable analysis by checking duplicates, identifying missing values, cleaning data types and text values, and performing necessary preprocessing.

### **6A. DATA CLEANING & PREPROCESSING**

This section is intentionally added before EDA/analysis so the notebook follows the standard workflow. Existing analysis cells are kept unchanged.

```python
# ------------------------------------------------------------
# 6A. DATA CLEANING & PREPROCESSING
# ------------------------------------------------------------
# Minimal preprocessing added before EDA.
# Existing analysis code is kept unchanged.

# Clean column names and text values
df.columns = df.columns.str.strip()

for column in df.select_dtypes(include=["object", "category"]).columns:
    df[column] = df[column].astype(str).str.strip()

# Handle missing values in predictor/input columns.
# Keep the target Yield_Tonnes_Ha missing because ML target rows
# are handled explicitly later before model training.
target_column = "Yield_Tonnes_Ha"

numeric_input_columns = [
    col for col in df.select_dtypes(include=np.number).columns
    if col != target_column
]

categorical_input_columns = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

for column in numeric_input_columns:
    if df[column].isnull().any():
        df[column] = df[column].fillna(df[column].median())

for column in categorical_input_columns:
    if df[column].isnull().any():
        mode_value = df[column].mode()
        if not mode_value.empty:
            df[column] = df[column].fillna(mode_value.iloc[0])

print("Data cleaning and preprocessing completed.")
print("Remaining missing values:")
display(df.isnull().sum()[df.isnull().sum() > 0].to_frame("Missing Values"))
print("Note: Missing Yield_Tonnes_Ha values are retained for later ML target handling.")
```

**6. CHECK MISSING VALUES**


```python
missing_values = df.isnull().sum()

missing_table = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": (missing_values / len(df)) * 100
})

display(missing_table.sort_values(
    by="Missing Values",
    ascending=False
))
```

**7. CHECK DUPLICATES**


```python
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

**8. REMOVE DUPLICATES**


```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
outlier_columns = [
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Revenue_INR",
    "Total_Cost_INR",
    "Profit_INR",
    "Water_Used_m3"
]

for column in outlier_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print(
        f"{column}: {len(outliers)} potential outliers"
    )
```

**44. STATISTICAL SUMMARY BY SEASON**

```python
statistical_summary = df.groupby("Season")[
    [
        "Yield_Tonnes_Ha",
        "Production_Tonnes",
        "Revenue_INR",
        "Profit_INR",
        "Water_Used_m3",
        "Disease_Pest_Risk_pct"
    ]
].agg([
    "mean",
    "median",
    "std",
    "min",
    "max"
])

display(statistical_summary.round(2))
```

**45. ANOVA - YIELD DIFFERENCE BETWEEN SEASONS**

```python
season_groups = [
    group["Yield_Tonnes_Ha"].dropna().values
    for name, group in df.groupby("Season")
]

anova_result = stats.f_oneway(*season_groups)

print("ANOVA Result")
print("F-statistic:", anova_result.statistic)
print("P-value:", anova_result.pvalue)

if anova_result.pvalue < 0.05:
    print(
        "Conclusion: There is a statistically significant "
        "difference in average yield between at least some seasons."
    )
else:
    print(
        "Conclusion: No statistically significant difference "
        "in average yield was detected between seasons."
    )
```

**46. ANOVA - PROFIT DIFFERENCE BETWEEN SEASONS**

```python
profit_groups = [
    group["Profit_INR"].dropna().values
    for name, group in df.groupby("Season")
]

profit_anova = stats.f_oneway(*profit_groups)

print("Profit ANOVA Result")
print("F-statistic:", profit_anova.statistic)
print("P-value:", profit_anova.pvalue)

if profit_anova.pvalue < 0.05:
    print(
        "Conclusion: Average profit differs significantly "
        "between at least some seasons."
    )
else:
    print(
        "Conclusion: No statistically significant difference "
        "in average profit was detected between seasons."
    )
```

# **11. Machine Learning — Performance Classification**

**64. SELECT TARGET VARIABLE**

Here we predict Yield_Tonnes_Ha.

```python
target = "Yield_Tonnes_Ha"

print("Target variable:", target)
print("Target description: Crop Yield in Tonnes per Hectare")
```

**65. SELECT FEATURES**

⚠️ We should not use variables such as Production_Tonnes, Revenue_INR, or Profit_INR as input features when predicting yield because they are outcomes related to yield and can cause data leakage.

```python
features = [
    "Season",
    "Crop",
    "State",
    "District",
    "Farm_Area_Hectares",
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_pH",
    "Soil_Moisture_pct",
    "Nitrogen_kg_ha",
    "Phosphorus_kg_ha",
    "Potassium_kg_ha",
    "Fertilizer_kg_ha",
    "Pesticide_Litre_ha",
    "Irrigation_Method",
    "Seed_Quality_Score",
    "Water_Used_m3",
    "Disease_Pest_Risk_pct"
]

# Drop rows where the target variable has missing values
df_cleaned = df.dropna(subset=[target])

X = df_cleaned[features]
y = df_cleaned[target]

print("Features:", X.shape)
print("Target:", y.shape)
```

**Identify Categorical and Numerical Features**

**66. IDENTIFY FEATURE TYPES**

```python
# ------------------------------------------------------------
# 66. IDENTIFY FEATURE TYPES
# ------------------------------------------------------------

categorical_features = [
    "Season",
    "Crop",
    "State",
    "District",
    "Irrigation_Method"
]

numerical_features = [
    column for column in features
    if column not in categorical_features
]

print("Categorical Features:")
print(categorical_features)

print("\nNumerical Features:")
print(numerical_features)
```

**67. TRAIN-TEST SPLIT**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
```

**68. DATA PREPROCESSING**

```python
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(
            handle_unknown="ignore"
        ))
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numerical_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)

print("Preprocessing pipeline created.")
```

**69. LINEAR REGRESSION MODEL**

```python
train_data = X_train.copy()
train_data[target] = y_train

test_data = X_test.copy()
test_data[target] = y_test

train_data = train_data.dropna(subset=[target])
test_data = test_data.dropna(subset=[target])

X_train = train_data.drop(columns=[target])
y_train = train_data[target]

X_test = test_data.drop(columns=[target])
y_test = test_data[target]

# Linear Regression Model
linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)

# Train model
linear_model.fit(X_train, y_train)

# Prediction
linear_predictions = linear_model.predict(X_test)

# Evaluation
linear_mae = mean_absolute_error(
    y_test,
    linear_predictions
)

linear_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        linear_predictions
    )
)

linear_r2 = r2_score(
    y_test,
    linear_predictions
)

print("LINEAR REGRESSION RESULTS")
print("=" * 40)
print("MAE :", round(linear_mae, 4))
print("RMSE:", round(linear_rmse, 4))
print("R²  :", round(linear_r2, 4))
```

**70. RANDOM FOREST REGRESSION**

This is the main ML model

```python
random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)

random_forest_model.fit(
    X_train,
    y_train
)

rf_predictions = random_forest_model.predict(
    X_test
)

rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_predictions
    )
)

rf_r2 = r2_score(
    y_test,
    rf_predictions
)

print("RANDOM FOREST RESULTS")
print("=" * 40)
print("MAE :", round(rf_mae, 4))
print("RMSE:", round(rf_rmse, 4))
print("R²  :", round(rf_r2, 4))
```

**71. GRADIENT BOOSTING REGRESSION**

```python
gradient_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            GradientBoostingRegressor(
                n_estimators=150,
                random_state=42
            )
        )
    ]
)

gradient_model.fit(
    X_train,
    y_train
)

gradient_predictions = gradient_model.predict(
    X_test
)

gradient_mae = mean_absolute_error(
    y_test,
    gradient_predictions
)

gradient_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        gradient_predictions
    )
)

gradient_r2 = r2_score(
    y_test,
    gradient_predictions
)

print("GRADIENT BOOSTING RESULTS")
print("=" * 40)
print("MAE :", round(gradient_mae, 4))
print("RMSE:", round(gradient_rmse, 4))
print("R²  :", round(gradient_r2, 4))
```

StandardScaler is added for the numerical variables. Categorical variables continue to use one-hot encoding. Tree-based models do not require scaling, while scaling is useful for the Linear Regression extension.

```python
# ------------------------------------------------------------
# 100. FEATURE SCALING
# ------------------------------------------------------------

from sklearn.preprocessing import StandardScaler

scaled_numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

scaled_preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            scaled_numeric_transformer,
            numerical_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)

print("StandardScaler added for numerical ML features.")
```

**101. SCALED LINEAR REGRESSION**

A scaled Linear Regression model is added as a preprocessing comparison. The original Linear Regression, Random Forest and Gradient Boosting models are not changed.

```python
scaled_linear_model = Pipeline(
    steps=[
        ("preprocessor", scaled_preprocessor),
        ("model", LinearRegression())
    ]
)

scaled_linear_model.fit(X_train, y_train)

scaled_linear_predictions = scaled_linear_model.predict(X_test)

scaled_linear_mae = mean_absolute_error(
    y_test,
    scaled_linear_predictions
)

scaled_linear_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        scaled_linear_predictions
    )
)

scaled_linear_r2 = r2_score(
    y_test,
    scaled_linear_predictions
)

print("SCALED LINEAR REGRESSION RESULTS")
print("=" * 45)
print("MAE :", round(scaled_linear_mae, 4))
print("RMSE:", round(scaled_linear_rmse, 4))
print("R²  :", round(scaled_linear_r2, 4))
```

# **14. Agriculture Performance Dashboard**
### **Purpose**
Present the most important agricultural KPIs, seasonal comparisons, crop and regional performance, resource usage, risk indicators, and machine learning results in a single visual dashboard.

**104. AGRICULTURE PERFORMANCE DASHBOARD**

This dashboard is an added presentation layer. The existing analytics and machine-learning cells are not changed. It summarizes the most important agricultural KPIs, seasonal performance, crop-season performance, regional performance, resource usage, risk, and ML results.

```python
import matplotlib.pyplot as plt
import seaborn as sns

def show_agriculture_dashboard(
    season="All",
    crop="All",
    state="All",
    irrigation_method="All"
):
    data = df.copy()

    # Dashboard filters
    if season != "All":
        data = data[data["Season"] == season]
    if crop != "All":
        data = data[data["Crop"] == crop]
    if state != "All":
        data = data[data["State"] == state]
    if irrigation_method != "All":
        data = data[data["Irrigation_Method"] == irrigation_method]

    if data.empty:
        print("No records match the selected filters.")
        return

    # KPI calculations
    total_farms = data["Farm_ID"].nunique() if "Farm_ID" in data.columns else len(data)
    avg_yield = data["Yield_Tonnes_Ha"].mean()
    avg_profit = data["Profit_INR"].mean()
    avg_revenue = data["Revenue_INR"].mean()
    avg_water_eff = data["Water_Efficiency_t_per_1000m3"].mean()
    avg_risk = data["Disease_Pest_Risk_pct"].mean()

    fig = plt.figure(figsize=(20, 15))
    # Adjusted gridspec for 5 rows to accommodate 2 rows of KPIs
    gs = fig.add_gridspec(5, 3, height_ratios=[0.9, 0.7, 0.7, 1.45, 1.45])

    # Dashboard title
    ax_title = fig.add_subplot(gs[0, :])
    ax_title.axis("off")
    ax_title.text(0.5, 0.72, "SEASONAL AGRICULTURE PERFORMANCE DASHBOARD",
                  ha="center", va="center", fontsize=22, fontweight="bold")
    ax_title.text(0.5, 0.28,
                  f"Season: {season}   |   Crop: {crop}   |   State: {state}   |   Irrigation: {irrigation_method}",
                  ha="center", va="center", fontsize=11)

    # KPI cards - now spread across two rows (rows 1 and 2)
    kpis = [
        ("Total Farms", f"{total_farms:,}"),
        ("Avg Yield", f"{avg_yield:.2f} t/ha"),
        ("Avg Profit", f"₹{avg_profit:,.0f}"),
        ("Avg Revenue", f"₹{avg_revenue:,.0f}"),
        ("Water Efficiency", f"{avg_water_eff:.2f}"),
        ("Disease/Pest Risk", f"{avg_risk:.2f}%")
    ]

    for i, (label, value) in enumerate(kpis):
        row = 1 if i < 3 else 2  # First 3 KPIs in row 1, next 3 in row 2
        col_idx = i % 3
        ax = fig.add_subplot(gs[row, col_idx])
        ax.axis("off")
        ax.text(0.5, 0.62, value, ha="center", va="center",
                fontsize=19, fontweight="bold")
        ax.text(0.5, 0.28, label, ha="center", va="center", fontsize=10)
        for spine in ax.spines.values():
            spine.set_visible(True)

    # 1. Season-wise yield - shifted to row 3
    ax1 = fig.add_subplot(gs[3, 0])
    season_yield = data.groupby("Season", observed=True)["Yield_Tonnes_Ha"].mean().sort_values(ascending=False)
    sns.barplot(x=season_yield.index, y=season_yield.values, ax=ax1)
    ax1.set_title("Average Yield by Season")
    ax1.set_xlabel("Season")
    ax1.set_ylabel("Yield (Tonnes/Ha)")
    ax1.tick_params(axis="x", rotation=35)

    # 2. Season-wise profit - shifted to row 3
    ax2 = fig.add_subplot(gs[3, 1])
    season_profit = data.groupby("Season", observed=True)["Profit_INR"].mean().sort_values(ascending=False)
    sns.barplot(x=season_profit.index, y=season_profit.values, ax=ax2)
    ax2.set_title("Average Profit by Season")
    ax2.set_xlabel("Season")
    ax2.set_ylabel("Profit (INR)")
    ax2.tick_params(axis="x", rotation=35)

    # 3. Crop x season heatmap - shifted to row 3
    ax3 = fig.add_subplot(gs[3, 2])
    pivot = pd.pivot_table(
        data, values="Yield_Tonnes_Ha", index="Crop", columns="Season", aggfunc="mean"
    )
    # Use a light, readable colour scale so a single value is not rendered black.
    if pivot.empty:
        ax3.text(0.5, 0.5, "No crop-season data available",
                 ha="center", va="center", fontsize=11)
        ax3.set_xticks([])
        ax3.set_yticks([])
    else:
        vmin = float(pivot.min().min())
        vmax = float(pivot.max().max())
        # Avoid a single-colour map when only one value remains after filtering.
        if abs(vmax - vmin) < 1e-12:
            pad = max(abs(vmin) * 0.10, 0.10)
            vmin -= pad
            vmax += pad

        sns.heatmap(
            pivot,
            annot=True,
            fmt=".2f",
            ax=ax3,
            cmap="YlGnBu",
            vmin=vmin,
            vmax=vmax,
            cbar=True,
            linewidths=0.6,
            linecolor="white",
            annot_kws={"color": "black", "fontweight": "bold"}
        )

    ax3.set_title("Crop × Season Yield")
    ax3.set_xlabel("Season")
    ax3.set_ylabel("Crop")

    # 4. Top states by profit - shifted to row 4
    ax4 = fig.add_subplot(gs[4, 0])
    state_profit = data.groupby("State", observed=True)["Profit_INR"].mean().sort_values(ascending=False).head(10)
    sns.barplot(x=state_profit.values, y=state_profit.index, ax=ax4)
    ax4.set_title("Top 10 States by Average Profit")
    ax4.set_xlabel("Profit (INR)")
    ax4.set_ylabel("State")

    # 5. Water usage vs yield - shifted to row 4
    ax5 = fig.add_subplot(gs[4, 1])
    sns.scatterplot(
        data=data, x="Water_Used_m3", y="Yield_Tonnes_Ha",
        hue="Season", alpha=0.65, ax=ax5, legend=False
    )
    ax5.set_title("Water Usage vs Yield")
    ax5.set_xlabel("Water Used (m³)")
    ax5.set_ylabel("Yield (Tonnes/Ha)")

    # 6. Disease/pest risk by season - shifted to row 4
    ax6 = fig.add_subplot(gs[4, 2])
    risk = data.groupby("Season", observed=True)["Disease_Pest_Risk_pct"].mean().sort_values(ascending=False)
    sns.barplot(x=risk.index, y=risk.values, ax=ax6)
    ax6.set_title("Average Disease/Pest Risk by Season")
    ax6.set_xlabel("Season")
    ax6.set_ylabel("Risk (%)")
    ax6.tick_params(axis="x", rotation=35)

    plt.tight_layout()
    plt.show()

# Default dashboard view
show_agriculture_dashboard()
```
